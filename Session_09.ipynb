Use a for loop and an accumulator named total to calculate the sum from 1 through n.

### Video Explanation

## Question 3: Compute Factorial Iteratively

Use an accumulator and a for loop to calculate the factorial of the supplied number.

### Video Explanation
YouTube Link:

```python
number = 6 
factorial = 1
print("Number:", number)
for value in range(1, number + 1):
    factorial = factorial * value
    print("factorial after", value, ":", factorial)

print("Final Factorial:", factorial)
```

## Question 5: Find the First Matching List Value

Use a for loop and break to find the first value equal to the target.

### Video Explanation
YouTube Link:

```python
values = [12,7,18,5,21]
target = 18
found_value = None
print ("Values:", values)
print("Target:", target)

for value in values:
    if value == target:
        found_value = value 
        break

print("Found Values:", found_value)
```

## Question 6: Read Values Until a Sentinel

Process the entries until 'done' is reached, convert the earlier text values to integers, and add them to total.

### Video Explanation
YouTube Link:

```python
entries = ["12", "7", "5", "done"]
total = 0
print("Enteries:", entries) 
for entry in entries:
    if entry == "done":
        break
    value = int(entry)
    total = total + value
    print("Running Total:", total)

print("Final Total:", total)
```

```python
entries = ["done"]
total = 0
print("Enteries:", entries) 
for entry in entries:
    if entry == "done":
        break
    value = int(entry)
    total = total + value
    print("Running Total:", total)

print("Final Total:", total)
```
