## Question 6: Classify a Triangle from Side Lengths

First validate the triangle inequality, then classify a valid triangle.

### Video Explanation
YouTube Link:

```python
side_1 = 5
side_2 = 5
side_3 = 8
is_valid_triangle = (side_1 + side_2 > side_3 and side_1 + side_3 > side_2 and side_2 + side_3 > side_1)
if not is_valid_triangle:
    classification = "invalid triangle"
elif side_1 == side_2 == side_3:
    classification = "equilateral"
elif side_1 == side_2 or side_1 == side_3 or side_2 == side_3:
    classification = "isosceles"
else: 
    classification = "scalene"
print("Sides:", side_1, side_2, side_3)
print("Valid Triangle:", is_valid_triangle)
print("Classifciation:", classification)
```

## Question 7: Select a Weekday with match and case

Use match and case to convert a day number from 1 through 7 into its weekday name.

### Video Explanation
YouTube Link:

```python
day_number = 3
match day_number:
    case 1:
        day_name = "Monday"
    case 2: 
        day_name = "Tuesday"
    case 3:
        day_name = "Wednesday"
    case 4:
        day_name = "Thursday"
    case 5:
        day_name = "Friday"
    case 6 :
        day_name = "Saturday"
    case 7:
        day_name = "Sunday"
    case _:
        day_name = "Invalid day number"

print("Day Number:", day_number)
print("Day Name:", day_name)
```

## Question 8: Match Transaction Records with a Guard

Match a transaction tuple and use a guard to flag large withdrawals for review.

### Video Explanation
YouTube Link:

```python
transaction = ("Withdrawal", 25000)
match transaction:
    case("Deposit", amount) if amount > 0:
        result = "Deposit accepted"
    case("withdrawal", amount) if amount > 20000:
        result = "withdrawal flagged for review"
    case("wwithdrawal", amount) if amount > 0:
        result = "Withdrawal accepted"
    case _:
        result = "Unsupported transaction"

print("Transaction:", transaction)
print("Result:", result)
```
