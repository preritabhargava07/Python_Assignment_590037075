## Question 2: Stor Quotient and Remainder in a Tuple
Calculate Quotient and Remainder, store them in a tuple, and unpack the tuple again.

### Video Explaination
Youtube Link:

```python
dividend = 1137
divisor = 60
quotient = dividend // divisor
remainder = dividend % divisor
result = (quotient, remainder)
unpacked_quotient, unpacked_remainder = result
print("Dividend:", dividend)
print("Divisor:", divisor)
print("Result:", result)
print("Unpacked quotient:", unpacked_quotient)
print("Unpacked remainder:", unpacked_remainder)
```

## Question 4: Copy text after Converting it to uppercase 
Write source text, read it, convert it to uppercase, write it to another file, and verify the results.

### Video Explanation
Youtube Link:

```python
from pathlib import Path
source_text = "Python files are readable"

source_path = Path("source.txt")
destination_path = Path("destination.txt")

source_path.write_text(source_text, encoding="utf-8")
read_text = source_path.read_text(encoding="utf-8")

uppercase_text = read_text.upper()
destination_path.write_text(uppercase_text, encoding="utf-8")

verified_text = destination_path.read_text(encoding="utf-8")

print("Source Text:", source_text)
print("Uppercase Tect:", uppercase_text)
print("Verified Text:", verified_text)

source_path.unlink()
destination_path.unlink()
```

## Question 5: Read three known lines explicitly 
Write three lines, read all lines, and retrieve the first three using indexes.

### Video Explanation
Youtube Link:

```python
from pathlib import Path
line_1 = "North"
line_2 = "South"
line_3 = "East"

path = Path("three_lines.txt")
file_text = line_1 + "\n" + line_2 + "\n" + line_3 + "\n"
path.write_text(file_text, encoding="utf-8")

lines = path.read_text(encoding="utf-8").splitlines()

first_line = lines[0]
second_line = lines[1]
third_line = lines[2]

print("All lines:", lines)
print("First Line:", first_line)
print("Second Line:", second_line)
print("Third Line:", third_line)
```

## Question 7: Write CSV - Style recoeds without a loop 
Join the header and two rows using newline characters, write the complete text once, read it back, and delete the file 

### Video Explanation
Youtube Link:

```python
from pathlib import Path
header = "name,score"
row_1 = "Asha, 82"
row_2 = "Ravi, 76"
csv_text = header + "\n" + row_1 + "\n" + row_2

path = Path("record.csv")
path.write_text(csv_text, encoding="utf-8")

read_text = path.read_text(encoding="utf-8")

print("CSV Text:")
print(csv_text)
print("Read Back;")
print(read_text)
path.unlink()
```

## Question 8: Encode and Decode Unicode Text with UTF - 8
Encode the message into UTF - 8 bytes, decode it back to text, and verify equality with the original.

### Video explanation
Youtube Link:

```python
message = " Namaste Python"
encoded_message = message.encode("utf-8")
decoded_message = encode_message.decode("utf-8")

print("Original Message:", message)
print("Encoded Message:", encoded_message)
print("Decoded Message:", decoded_message)
print("Matches original:", decoded_message == message)
```
