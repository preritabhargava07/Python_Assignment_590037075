## Question 1: Count Vowels 
Count the vowels in the given text without making the count case - sensitive.

## Question 3: Check a Palindrome

Normalize the supplied text and compare it with its reverse to check whether it is a palindrome.

### Video Explanation
Youtube Link:

```python
text = "Never odd or even"
normalized_text = text.strip().lower().replace(" ", "")
reversed_text = normalized_text == reversed_text
is_palindrome = normalized_text == reversed_text
print("Normalized text:", normalized_text)
print("Reversed Text:", reversed_text)
print("Is Palindrome:", is_palindrome)
```

```python
text = "Python"
normalized_text = text.strip().lower().replace(" ", "")
reversed_text = normalized_text == reversed_text
is_palindrome = normalized_text == reversed_text
print("Normalized text:", normalized_text)
print("Reversed Text:", reversed_text)
print("Is Palindrome:", is_palindrome)
```

## Question 8: Format a Receipt 

Calculate the total price and display a formatted receipt using an f - string.

### Video Explantion
Youtube Link:

```python
item = "Notebook"
quantity = 3 
unit_price = 125.5 
total_price = quantity * unit_price 
receipt = f"{item}:{quantity} x {unit_price:.2f} = {total_price:.2f}"
print(receipt)
```
