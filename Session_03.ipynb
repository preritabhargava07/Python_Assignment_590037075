## Question 1: Count Vowels Case-Insensitively

Convert the given text to lowercase and count the total number of vowels using count() for a, e, i, o, and u.

```python
text = "Data Science"
lowercase_text = text.lower()
vowel_count=(lowercase_text.count("a") + lowercase_text.count("e") + lowercase_text.count("i") + lowercase_text.count("o") + lowercase_text.count("u"))
print ("Original text:", text)
print("Lowercase text:", lowercase_text)
print("Vowel count:", vowel_count)
```

## Question 3: Test a Normalized Palindrome

Normalize the given text using strip(), lower(), and replace(), then compare it with its reversed form to test whether it is a palindrome.

```python
text = "Never Odd Or Even"
normalized_text = text.strip() .lower() .replace(" ","")
reversed_text = normalized_text[::-1]
is_palindrome = normalized_text ==reversed_text
print("Original text:", text)
print ("Normalized text:", normalized_text)
print("Reversed text:", reversed_text)
print("Palindrome:", is_palindrome)
```

```python
text = "Python"
normalized_text = text.strip() .lower() .replace(" ","")
reversed_text = normalized_text[::-1]
is_palindrome = normalized_text ==reversed_text
print("Original text:", text)
print ("Normalized text:", normalized_text)
print("Reversed text:", reversed_text)
print("Palindrome:", is_palindrome)
```

## Question 4: Extract a Filename Extension

Use rsplit(".", 1) to split the filename at its final period and extract the extension by indexing the resulting pieces.

```python
filename = "quarterly_report.csv"
pieces = filename.rsplit(".",1)
extension = pieces[1]
print("Filename:",filename)
print("Pieces;",pieces)
print("Extension:", extension)
```

## Question 5: Replace Repeated Whitespace

Replace each group of three spaces in the given text with one space using replace().

```python
text = "Python   makes   text   readable"
repeated_space = "   "
cleaned_text = text.replace(repeated_space, " ")

print("Original text:", text)
print("Repeated value:",repr(repeated_space))
print("Cleaned text:",cleaned_text)
```

## Question 6: Create Initials From a Full Name

Split the full name into three parts, index the first character of each part, and concatenate the characters with periods to create initials.

```python
full_name = "Asha Mehta Singh"
name_parts =full_name.split()
initials = name_parts[0][0] + "." + name_parts[1][0] + "." + name_parts[2][0] + "."

print("Full name:", full_name)
print("Name parts:", name_parts)
print("Initials:", initials)
```

## Question 8: Format a Receipt Line With Two Decimals

Calculate the line total from quantity and unit price, then use one f-string to display the item, quantity, unit price, and total with two decimal places.

```python
item = "Notebook"
quantity = 3 
unit_price = 125.5
line_total = quantity * unit_price
receipt_line = f"Item:{item}, Quantity: {quantity}, Unit price; {unit_price:.2f}, Total: {line_total:.2f}"
print("Item:", item)
print("Quantity;",quantity)
print("Unit price;", unit_price)
print("Line total:", line_total)
print("Receipt:",receipt_line)
```
