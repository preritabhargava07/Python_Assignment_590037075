## Question 2: Build Lazy Even Squares 
Create a generator expression that produces the square of each even number.

### Video Explanation 
Youtube Link:


```python
numbers = range (1,11)
even_squares = (n * n for n in numbers if n % 2 ==0)
result = list(even_squares)
print("Numvers:", list(numbers))
print("Even Squares:", result)
```

```python
numbers = range (1,2)
even_squares = (n * n for n in numbers if n % 2 ==0)
result = list(even_squares)
print("Numvers:", list(numbers))
print("Even Squares:", result)
```

## Question 5: Filter Records with a Comprehension

Keep only records whose score is at least the minimum score.

### Video Explanation
YouTube Link:

```python
records = [("Asha", 82),("Ravi", 64),("Mira", 91)]
minimum_score = 75
selected_records = [record for record in records if record[1] >= minimum_score]
print("records:", records)
print("Minimum Score:", minimum_score)
print("Selected Records:", selected_records)
```

## Question 8: Combine zip with a Dictionary Comprehension

Pair products with prices using zip and create a dictionary from those pairs.

### Video Explanation
YouTube Link:

```python
products = ["Pen", "Notebook", "Marker"]
prices = [20,125,45]
price_by_product = {product: price for product, price in zip(products, prices)}
print("Products:", products)
print("Prices:", prices)
print("Price by Product:", price_by_product)
```

```python
products = ["Pen", "Notebook"]
prices = [20]
price_by_product = {product: price for product, price in zip(products, prices)}
print("Products:", products)
print("Prices:", prices)
print("Price by Product:", price_by_product)
```
