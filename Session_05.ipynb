### Video Explanation

## Question 4: Remove an Entry with pop()

## Question 6: Merge Two Dictionaries with the Union Operator

Merge default settings with user settings using the dictionary union operator.

### Video Explanation

```python
defaults = {"theme": "light", "page_size": 20}
user = {"theme": "dark"}
settings = defaults | user
print("Defaults:", defaults)
print("User:", user)
print("Settings:", settings)
```

## Question 7: Inspect Keys, Values, and Items Views

Create and display the keys, values, and items views of the dictionary.

### Video Explanation
YouTube Link:

```python
inventory = {"pen": 40, "Notebook": 25, "Marker": 18}
keys_view = inventory.keys()
values_view = inventory.values()
items_view = inventory.items()

print("Inventory:", inventory)
print("Keys View:")
print("Values View:", values_view)
print("Items View:", items_view)
```

## Question 8: Compare a Dictionary Alias with a Copy

Create an alias and a copy, update the original dictionary, and compare all three.

### Video Explanation
YouTube Link:

```python
original = {"theme": "light", "page_size": 20 }
alias = original
copied = original.copy()
print("Before Update:")
print("Original:", original)
print("Alias:", alias)
print("Copied:",copied)
original["theme"] = "dark"
print("After Update:")
print("Original:", original)
print("Alias:", alias)
print("Copied:", copied)
```

```python
original = {}
alias = original
copied = original.copy()
print("Before Update:")
print("Original:", original)
print("Alias:", alias)
print("Copied:",copied)
original["theme"] = "dark"
print("After Update:")
print("Original:", original)
print("Alias:", alias)
print("Copied:", copied)
```
