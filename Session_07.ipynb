Compare equal list values, distinct list objects, and an alias.

## Question 6: Capture a Computed Value with the Walrus Operator

Use the walrus operator to store the largest value, then compare it with 10.

### Video Explanation
YouTube Link:

```python
numbers = [3, 8, 12]
largest = None
is_large = (largest := max(numbers)) > 10
print("numbers:", numbers)
print("Largest:", largest)
print("Largest>10:", is_large)
```

```python
numbers = [-8, -3, -12]
largest = None
is_large = (largest := max(numbers)) > 10
print("numbers:", numbers)
print("Largest:", largest)
print("Largest>10:", is_large)
```
