## Question 5: Convert Seconds into Hours, Minutes and Seconds

Convert the given total number of seconds into hours, minutes and seconds 
### Video Explanation
Youtube Link

```python
total_seconds = 7384
hours = total_seconds // 3600
remaining_seconds = total_seconds % 3600
minutes = remaining_seconds // 60
seconds = remaining_seconds % 60
print("Hours;", hours)
print("Minutes;", minutes)
print("Seconds:", seconds)
```
