
## Question 1: Compute Quotient and Remainder

Calculate the quotient and remainder using floor division and modulus, and check wheter the division is exact
### Video Explanation
Youtube Link:

```python
total_minutes = 137
minutes_per_hour = 60
quotient = total_minutes // minutes_per_hour
remainder = total_minutes % minutes_per_hour
reconstruction_check = quotient * minutes_per_hour + remainder == total_minutes
print("Quotient:", quotient)
print("Remainder:", remainder)
print("Reconstruction Check:", reconstruction_check)
```

```python
total_minutes = 60
minutes_per_hour = 60
quotient = total_minutes // minutes_per_hour
remainder = total_minutes % minutes_per_hour
reconstruction_check = quotient * minutes_per_hour + remainder == total_minutes
print("Quotient:", quotient)
print("Remainder:", remainder)
print("Reconstruction Check:", reconstruction_check)
```

## Question 4: Calculate Compound Growth

Calculate the final amount after compound growth and the total growth over the period.

### Video Explanation
Youtube Link:

```python
principal = 50000
annual_rate = 0.08
years = 2 
final_amount = principal * (1+ annual_rate) ** years
growth = final_amount - principal
print("Final Amount:", final_amount)
print ("Growth:", growth)
```
