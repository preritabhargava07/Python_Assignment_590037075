
## Question 1: Compute Quotient and Remainder

Starting with the given dividend and divisor, calculate the quotient using floor division and the remainder using modulus. Verify that:

dividend = quotient × divisor + remainder

```python
dividend = 137
divisor = 60
quotient = dividend // divisor
remainder = dividend % divisor
verification = dividend == quotient * divisor + remainder
print("Dividend:", dividend)
print("Divisor:", divisor)
print("Quotient:", quotient)
print("Remainder:", remainder)
print("Verified:", verification)
```

## Question 4: Calculate Compound Growth

Starting with the given principal, annual rate, and number of years, calculate the final amount using compound growth and then calculate the growth amount.

```python
principal = 50000.0
annual_rate = 0.08
years = 2
final_amount = principal * (1 + annual_rate) ** years
growth_amount = final_amount - principal
print("Principal:",principal)
print("Annual rate:", annual_rate)
print("Years:",years)
print("Final amount:", final_amount)
print("Growth amount:", growth_amount)
```
