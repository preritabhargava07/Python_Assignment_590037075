## Question 2: Compare append() with extend()

Make two copies of the supplied list. Use append() to add new_items as one nested item and extend() to add the individual items, then compare the results.

```python
items = ["Pen","Notebook"]
new_items = ["Marker","Folder"]

appended_items = items.copy()
appended_items.append(new_items)

extended_items = items.copy()
extended_items.extend(new_items)

print("Original items:", items)
print("New items;", new_items)
print("After append:", appended_items)
print("After extend:", extended_items)
```

```python
items = []
new_items = ["Marker","Folder"]

appended_items = items.copy()
appended_items.append(new_items)

extended_items = items.copy()
extended_items.extend(new_items)

print("Original items:", items)
print("New items;", new_items)
print("After append:", appended_items)
print("After extend:", extended_items)
```

## Question 3: Use insert(), remove(), and pop()

Insert Chennai at index 1, remove the supplied city by value, then pop and store the final city. Display the list after each change.

```python
cities = ["Delhi", "Mumbai", "Pune", "Jaipur"]
removed_city = "Pune"
print("Starting Cities:", cities)

cities.insert(1,"Chennai")
print("After insert:", cities)

cities.remove(removed_city)
print("After emove:", cities)

popped_city = cities.pop()
print("Popped city:", popped_city)
print("Final cities:", cities)
```

## Question 4: Compare sort() with sorted()

Use sorted() to create a new ascending list without changing the original, then use sort(reverse=True) to sort the original list in descending order.

```python
scores = [72,91,68,85,77]

print("Starting scores:", scores)
sorted_scores = sorted(scores)
print ("Sorted copy:", sorted_scores)
print("Original after sorted():", scores)
scores.sort(reverse=True)
print("Original after sort(reverse=True):", scores)
print("Sorted copy:", sorted_scores)
```

## Question 6: Demonstrate List Aliasing

Create an alias for the original list, append 40 through the alias, and show that the original list also changes.

```python
original = [10,20,30]
alias = original
print("Original before append:", original)
print("Alias before append:", alias)
alias.append(40)
print("Original after append:", original)
print("Alias after append:", alias)
print("Same list:", alias is original)
```

```python
original = []
alias = original
print("Original before append:", original)
print("Alias before append:", alias)
alias.append(40)
print("Original after append:", original)
print("Alias after append:", alias)
print("Same list:", alias is original)
```

## Question 7: Make an Explicit Independent Nested Copy

Copy each inner list separately to create an independent nested copy, then mutate the original and compare both nested lists.

```python
original = [[1],[2]]

independent = [original[0].copy(), original[1].copy()]

print("Original before mutation:",original)
print("Independent before mutation:", independent)

original[0].append(9)

print("Original after mutation;", original)
print("Independent after mutation:", independent)
```
