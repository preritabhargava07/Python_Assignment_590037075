## Question 6: Demonstrate List Aliasing

Create an alias, mutate the list through the alias, and show that the original also changes.

### Video Explanation
YouTube Link:

```python
original = [10, 20, 30]
alias = original
print("Before Mutation - Original:", original)
print("Before Mutation - Alias:", alias)
alias.append(40)
print("After Mutation - Original:", original)
print("After Mutation - Alias:", alias)
print("Same object:", alias is original)
```

```python
original = []
alias = original
print("Before Mutation - Original:", original)
print("Before Mutation - Alias:", alias)
alias.append(40)
print("After Mutation - Original:", original)
print("After Mutation - Alias:", alias)
print("Same object:", alias is original)
```
